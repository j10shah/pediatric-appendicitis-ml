```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("logs/class_Dia_gb_log.csv")

plt.figure()


for k in sorted(df["k"].unique()):
    fold_df = df[df["k"] == k]

    plt.plot(fold_df["epoch"], fold_df["train_loss"], linestyle="-", label=f"train k={k}")
    plt.plot(fold_df["epoch"], fold_df["val_loss"], linestyle="--", label=f"val k={k}")

avg_df = df.groupby("epoch")[["train_loss", "val_loss"]].mean().reset_index()
plt.plot(avg_df["epoch"], avg_df["train_loss"], linewidth=3, label="avg train")
plt.plot(avg_df["epoch"], avg_df["val_loss"], linewidth=3, label="avg val")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("K-Fold Train/Validation Loss Curves")
plt.legend()
plt.show()
```
